```python
import os # For navigating through folders
import matplotlib.pyplot as plt # For drawing the EDA scatter plot
from PIL import Image # For opening and resizing images
from collections import Counter # To count overlapping image sizes
```

```python
base_dir = r'C:\Users\vinuga janandith\Desktop\AIML project\data'

widths = []
heights = []
resolution_counter = Counter()

print("Scanning images to analyze original dimensions...")

# Loop through all files
for root, dirs, files in os.walk(base_dir):
    if not dirs: 
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                file_path = os.path.join(root, file)
                try:
                    with Image.open(file_path) as img:
                        w, h = img.width, img.height
                        widths.append(w)
                        heights.append(h)
                        # Count how many images have this exact width/height
                        resolution_counter[(w, h)] += 1
                except Exception:
                    pass

# Print a breakdown of the resolutions found
print("-" * 40)
print(f"Total images scanned: {len(widths)}")
print("Resolution Breakdown:")
for res, count in resolution_counter.items():
    print(f" - {res[0]}x{res[1]} pixels: {count} images")
print("-" * 40)
```

```python
# EXPLORATORY DATA ANALYSIS (EDA)

plt.figure(figsize=(8, 6))
plt.scatter(widths, heights, color='purple', alpha=0.5, edgecolors='black', s=50)

plt.title('EDA: Original Image Resolutions', fontsize=14)
plt.xlabel('Image Width (Pixels)', fontsize=12)
plt.ylabel('Image Height (Pixels)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
