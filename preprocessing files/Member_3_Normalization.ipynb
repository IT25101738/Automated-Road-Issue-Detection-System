```python
import os # For finding folders
import numpy as np # For math and arrays
import matplotlib.pyplot as plt # For the EDA histogram
import matplotlib.image as mpimg # To open images
```

```python
base_dir = r'C:\Users\vinuga janandith\Desktop\AIML project\data'
mean_intensities = []

print("Normalizing pixels across ALL images...")

# Loop through all files
for root, dirs, files in os.walk(base_dir):
    if not dirs: 
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                file_path = os.path.join(root, file)
                try:
                    img = mpimg.imread(file_path)
                    if img is not None:
                        # DATA PREPROCESSING: Normalization
                        # Scale pixels from 0-255 down to 0-1 range
                        normalized_img = img / 255.0
                        
                        # Save the average brightness of this normalized image
                        mean_intensities.append(np.mean(normalized_img))
                except Exception:
                    pass

print(f"Successfully normalized and analyzed {len(mean_intensities)} images.")
```

```python
# EXPLORATORY DATA ANALYSIS (EDA)
# Plot a Histogram of the pixel intensities
plt.figure(figsize=(10, 5))
plt.hist(mean_intensities, bins=50, color='teal', edgecolor='black', alpha=0.7)

plt.title('EDA: Normalized Pixel Intensity Distribution', fontsize=14)
plt.xlabel('Mean Pixel Intensity (Scaled 0 to 1)', fontsize=12)
plt.ylabel('Number of Images', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
